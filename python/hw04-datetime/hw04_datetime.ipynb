## Задание 1

Печатные газеты использовали свой формат дат для каждого выпуска. Для каждой газеты из списка напишите формат указанной даты для перевода в объект `datetime`:

- The Moscow Times — `Wednesday, October 2, 2002`
- The Guardian — `Friday, 11.10.13`
- Daily News — `Thursday, 18 August 1977`

Программа должна выводить на экран объекты типа `datetime`, соответствующие датам в условии.

### Решение

````python
from datetime import datetime, timedelta


def parse_newspaper_date(date_text, date_format):
    """Перевести строку даты в datetime по заданному формату."""
    return datetime.strptime(date_text, date_format)


def newspaper_date_error(date_text, date_format):
    """Проверка: строка и формат непустые, дата парсится."""
    if not isinstance(date_text, str) or not date_text.strip():
        return "Дата должна быть непустой строкой."
    if not isinstance(date_format, str) or not date_format.strip():
        return "Формат должен быть непустой строкой."
    try:
        parse_newspaper_date(date_text, date_format)
    except ValueError:
        return "Дата не соответствует формату."
    return None


def print_parsed_date(newspaper, date_text, date_format):
    error = newspaper_date_error(date_text, date_format)
    if error:
        print(f"{newspaper}: {error}")
    else:
        print(f"{newspaper}: {parse_newspaper_date(date_text, date_format)}")
````

````python
# @title The Moscow Times
moscow_date = "Wednesday, October 2, 2002"  # @param {type:"string"}
moscow_format = "%A, %B %d, %Y"  # @param {type:"string"}

print_parsed_date("The Moscow Times", moscow_date, moscow_format)
````

````python
# @title The Guardian
guardian_date = "Friday, 11.10.13"  # @param {type:"string"}
guardian_format = "%A, %d.%m.%y"  # @param {type:"string"}

print_parsed_date("The Guardian", guardian_date, guardian_format)
````

````python
# @title Daily News
daily_date = "Thursday, 18 August 1977"  # @param {type:"string"}
daily_format = "%A, %d %B %Y"  # @param {type:"string"}

print_parsed_date("Daily News", daily_date, daily_format)
````

````python
# @title Любая другая дата
custom_name = "Custom"  # @param {type:"string"}
custom_date = ""  # @param {type:"string"}
custom_format = ""  # @param {type:"string"}

print_parsed_date(custom_name, custom_date, custom_format)
````

## Задание 2 (необязательное)

Напишите функцию `date_range`, которая возвращает список дат за период от `start_date` до `end_date`. Даты вводятся в формате `YYYY-MM-DD`.

При неверном формате или если `start_date > end_date` возвращается пустой список.

Примеры:

```
date_range('2022-01-01', '2022-01-03')
['2022-01-01', '2022-01-02', '2022-01-03']

date_range('2022-01-03', '2022-01-01')
[]

date_range('2022-02-30', '2022-02-31')
[]
```

### Решение

````python
def date_range(start_date, end_date):
    """Список дат YYYY-MM-DD от start_date до end_date включительно.
    При ошибке формата или start_date > end_date возвращает [].
    """
    try:
        start = datetime.strptime(start_date, "%Y-%m-%d")
        end = datetime.strptime(end_date, "%Y-%m-%d")
    except (TypeError, ValueError):
        return []

    if start > end:
        return []

    dates = []
    current = start
    while current <= end:
        dates.append(current.strftime("%Y-%m-%d"))
        current += timedelta(days=1)
    return dates
````

````python
# @title Период
start_date = "2022-01-01"  # @param {type:"string"}
end_date = "2022-01-03"  # @param {type:"string"}

print(date_range(start_date, end_date))
````
