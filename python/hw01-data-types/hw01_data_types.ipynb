## Задание 1

Дана переменная, в которой хранится слово из латинских букв. Напишите код, который выводит на экран:

- среднюю букву, если число букв в слове нечётное;
- две средних буквы, если число букв чётное.

Примеры:

1. `word = 'test'` → `es`
2. `word = 'testing'` → `t`

### Решение

````python
def latin_word_error(word):
    """Проверка: непустая строка только из латинских букв."""
    if not word:
        return "Слово пустое."
    if any(char.isspace() for char in word):
        return "В слове не должно быть пробелов."
    if not word.isalpha():
        return "Допустимы только буквы, без цифр и знаков."
    if not word.isascii():
        return "Нужны латинские буквы (A-Z, a-z)."
    return None


def middle_letters(word):
    """Средняя буква, если длина нечётная; две средние — если чётная."""
    length = len(word)
    middle = length // 2
    if length % 2 == 1:
        return word[middle]
    return word[middle - 1 : middle + 1]
````

````python
# @title Слово
word = "test"  # @param {type:"string"}

error = latin_word_error(word)
if error:
    print(error)
else:
    print(middle_letters(word))
````

## Задание 2

Вы делаете MVP dating-сервиса. Есть список юношей и девушек.

Гипотеза: лучшие рекомендации получатся, если отсортировать имена по алфавиту и познакомить людей с одинаковыми индексами после сортировки. Никого не знакомим, если кто-то может остаться без пары.

Примеры:

```
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

Идеальные пары:
Alex и Emma
Arthur и Kate
John и Kira
Peter и Liza
Richard и Trisha
```

```
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

Внимание, кто-то может остаться без пары!
```

### Решение

````python
def parse_name_list(raw_text):
    """Строка 'Ann, Bob' -> список имён."""
    return [name.strip() for name in raw_text.split(",") if name.strip()]


def names_error(names, title):
    """Проверка: список не пустой, в именах только буквы."""
    if not names:
        return f"Список {title} пустой."
    for name in names:
        letters_only = name.replace("-", "").replace(" ", "")
        if not letters_only.isalpha():
            return f"Некорректное имя в {title}: {name}"
    return None


def match_pairs(boys, girls):
    """Пары после сортировки или None, если длины разные."""
    if len(boys) != len(girls):
        return None
    return list(zip(sorted(boys), sorted(girls)))


def print_matching_result(boys, girls):
    pairs = match_pairs(boys, girls)
    if pairs is None:
        print("Внимание, кто-то может остаться без пары!")
        return
    print("Идеальные пары:")
    for boy, girl in pairs:
        print(f"{boy} и {girl}")
````

````python
# @title Списки имён
boys_raw = "Peter, Alex, John, Arthur, Richard"  # @param {type:"string"}
girls_raw = "Kate, Liza, Kira, Emma, Trisha"  # @param {type:"string"}

boys = parse_name_list(boys_raw)
girls = parse_name_list(girls_raw)

error = names_error(boys, "boys") or names_error(girls, "girls")
if error:
    print(error)
else:
    print_matching_result(boys, girls)
````
