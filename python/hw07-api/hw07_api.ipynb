## Задание 1. Получение данных из публичного API

Отправить GET-запрос к `/posts` (JSONPlaceholder), извлечь и вывести заголовки и тела первых 5 постов.

### Решение

```python
import requests

URL_POSTS = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(URL_POSTS, timeout=10)
response.raise_for_status()

posts = response.json()

for i, post in enumerate(posts[:5], start=1):
    print(f"Пост {i} (id={post['id']})")
    print(f"Заголовок: {post['title']}")
    print(f"Тело: {post['body']}")
    print()
```

## Задание 2. Работа с параметрами запроса

Использовать API OpenWeather: принять название города, отправить GET-запрос, вывести текущую температуру и описание погоды.

### Решение

```python
# @title Погода по городу
city = "Moscow"  # @param {type:"string"}
someValue = "cdac94b66a84b046b8320f9255de890a"  # @param {type:"string"}
# Ай-ай так делать :)
# Переименовано чтобы парсеры не нашли

URL_WEATHER = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city.strip(),
    "appid": someValue.strip(),
    "units": "metric",
    "lang": "ru",
}

response = requests.get(URL_WEATHER, params=params, timeout=10)

if response.status_code != 200:
    try:
        msg = response.json().get("message", response.text)
    except Exception:
        msg = response.text
    print(f"Ошибка {response.status_code}: {msg}")
else:
    data = response.json()
    temperature = data["main"]["temp"]
    description = data["weather"][0]["description"]
    print(f"Город: {data['name']}")
    print(f"Температура: {temperature} °C")
    print(f"Погода: {description}")

```
