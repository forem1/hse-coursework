## Задание 1

Реализуйте функцию `sum_distance(from, to)`, которая суммирует все числа от значения `from` до величины `to` включительно.

Если первое число больше второго, поменяйте их местами.

### Решение

```python
def sum_distance_error(from_, to):
    """Проверка: оба аргумента — целые числа."""
    if type(from_) is not int or type(to) is not int:
        return "Нужны целые числа."
    return None


def sum_distance(from_, to):
    """Сумма всех чисел от from_ до to включительно.
    from — ключевое слово Python, поэтому параметр называется from_.
    """
    if from_ > to:
        from_, to = to, from_
    return sum(range(from_, to + 1))
```

```python
# @title Границы отрезка
from_ = 1  # @param {type:"integer"}
to = 5  # @param {type:"integer"}

error = sum_distance_error(from_, to)
if error:
    print(error)
else:
    print(sum_distance(from_, to))
```

## Задание 2

Реализуйте функцию `trim_and_repeat()`, которая принимает три параметра:

- строку;
- `offset` — сколько символов обрезать слева;
- `repetitions` — сколько раз повторить получившуюся строку.

По умолчанию `offset = 0`, `repetitions = 1`. Функция возвращает полученную строку.

### Решение

```python
def trim_and_repeat_error(text, offset, repetitions):
    """Проверка типов и того, что offset и repetitions неотрицательные."""
    if not isinstance(text, str):
        return "Первый аргумент должен быть строкой."
    if type(offset) is not int or type(repetitions) is not int:
        return "offset и repetitions должны быть целыми числами."
    if offset < 0:
        return "offset не может быть отрицательным."
    if repetitions < 0:
        return "repetitions не может быть отрицательным."
    return None


def trim_and_repeat(text, offset=0, repetitions=1):
    """Обрезать строку слева на offset и повторить repetitions раз."""
    return text[offset:] * repetitions
```

```python
# @title Строка и параметры
text = "hello"  # @param {type:"string"}
offset = 2  # @param {type:"integer"}
repetitions = 3  # @param {type:"integer"}

error = trim_and_repeat_error(text, offset, repetitions)
if error:
    print(error)
else:
    print(trim_and_repeat(text, offset, repetitions))
```
